# kb_fluct: Diffusion Interaction Parameter ($k_D$) Demo

This notebook demonstrates how to compute the diffusion interaction parameter ($k_D$) for binary and ternary solutions based on Kirkwood-Buff Integrals (KBI) using the theoretical framework from Yang et al. (2025).

```python
from kb_fluct import BinarySystem, TernarySystem
```

### 1. Binary System Example
For a simple binary solution (e.g., water and urea).

```python
rho_solvent = 33.3  # Number density of solvent
rho_solute = 1.5    # Number density of solute

# Hypothetical KBI values
G11 = -0.02
G22 = 0.5
G12 = -0.1
eta0 = 0.89  # Viscosity of pure solvent (required for approximation)

binary = BinarySystem(rho1=rho_solvent, rho2=rho_solute, 
                      G11=G11, G22=G22, G12=G12, eta0=eta0)

print(f"Thermodynamic Factor (Gamma): {binary.calculate_Gamma():.4f}")
print(f"Exact k_D: {binary.calculate_kD_exact():.4f}")
print(f"Approximate k_D: {binary.calculate_kD_approx():.4f}")
```

### 2. Ternary System Example
For a multicomponent solution (e.g., water, protein, and a co-solute like arginine).

```python
rho_solvent = 33.3
rho_protein = 0.01  # Dilute protein
rho_cosolute = 0.5

# Hypothetical KBI values (including cross-terms)
G11 = -0.02
G22 = 1.2
G33 = -0.05
G12 = -0.1
G13 = 0.05
G23 = 0.8
eta0 = 0.89

ternary = TernarySystem(rho1=rho_solvent, rho2=rho_protein, rho3=rho_cosolute,
                        G11=G11, G22=G22, G33=G33, 
                        G12=G12, G13=G13, G23=G23, eta0=eta0)

print(f"Exact k_D (Ternary): {ternary.calculate_kD_exact():.4f}")
print(f"Approximate k_D (Ternary): {ternary.calculate_kD_approx():.4f}")
```
